# Docker

We will be learning about what docker is, it's benefits and how to use it.

## What is Docker?

It is a open-source platform that used OS-level virtualization package, ship and run applications inside lightweight, isolated containers.

Is is a platform for building, running and shipping applications

**Explanation**:

Suppose you create an application which requires some specific techstack eg : react v2, python 3.34, etc and also requires some specific configurations. 

This app runs perfectly in you machine but when you try to run it on your friends machine or when you deploy it you are suddenly unable to run it or get so many errors. This might be because of software missmatch, configurion missmatch, different runtime environmetns, etc.

Docker solves this by creating a single package of your applications, all the dependencies of your application with their specific verisons. You can run this package on any machine that has docker. 

Example : A new person joings your teams, they can simple use docker to install all the dependencies and run the applicaiton inside an isolated enviromnet called **Container**.

**Container** allows different applciations with different dependencies to run simultaniously on a single machine, isolated within their own container, without messing with one another. Once we are done with an explanation we can remove the application and its dependencies all together without casuing any problems to other applications.

## Containers v/s Virtual Machines


**Virtual Machines (VMs)**:  VMs Are basically an abstraction of physical hardware (using Hypervisors like VM ware Virtual Box, etc)
- one must install everything like OS(pay for it), softwars, etc on each VM(basically setting up a newly built PC). One machine can have multile VMs, and each must be setup separately.

- They are resource intensive and each VM takes a part/slice of actual hardware which is allocated by the user

- Slower to start - just like starting a pc after shutting down - wait for OS, open specific tool/software, etc

**Containers:** Containers are isolated environment for running an appliation.

- They allow running multiple apps in isolation

- Buit over OS - meaning a single OS is shared among all the containers on that machine 

- Light weight - quick to startup and less resources used and does not reuqire allocating specific part of hardware.

## Docker Architecture 
- uses client-server arcihtecture.

- Client <---REST API---> Server(Docker Engine) 

- Docker Engine : takes care of building and running Docker Containers.
- a Container is a process, just like other processes running on the computer. 
- Just like we discussed all containers share the host OS, specifically the kernel of the host OS.
- This means, cuz of how the OS kernel is designed, Windows can run both windows and linux containers as it shares some parts of the kernel from linux, Linux can run only linux containers, and mac? It has it own kernel which does not support containerized applications and docker solves this by using a light weight Linux VM to run linux based containers.
